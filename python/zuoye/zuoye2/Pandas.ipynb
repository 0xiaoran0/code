### 2.1 Series：一维数据结构

**Series** 是 Pandas 中最基础的数据结构，可以看作是带标签（索引）的**一维数组**。  
它兼具了数组（Array）和字典（Dict）的特点，既可以通过位置访问，也可以通过标签访问元素。

#### Series 特点：

- 是一组数据（任意数据类型）与一组与之相关的数据标签（索引）的组合。
- 支持类似 NumPy 数组的切片、索引和运算操作。
- 支持灵活的自定义索引。

Series 是构建 DataFrame 的基础，掌握 Series 的基本使用对后续学习至关重要。

```python
# 创建一个简单的 Series
s = pd.Series([10, 20, 30, 40, 50])
print("默认索引的 Series：")
print(s)
print()

# 查看索引和值
print("索引：", s.index)
print("值：", s.values)
print()

# 自定义索引的 Series
s2 = pd.Series([100, 200, 300], index=["a", "b", "c"])
print("自定义索引的 Series：")
print(s2)
print()

# Series 常见操作：访问元素、切片
print("访问元素（索引'b'对应的值）：", s2["b"])
print("\n切片（从索引'a'到索引'c'）：")
print(s2["a":"c"])
```

### 2.2 DataFrame：二维表格结构

**DataFrame** 是 Pandas 中最重要、最常用的数据结构之一，专门用于处理**二维表格型数据**。  
可以把 DataFrame 理解为带行索引和列索引的“表格”，类似于 Excel 表或数据库中的数据表。每一列可以看作是一个 Series，每一行则代表一条记录。

#### DataFrame 特点：

- 由按列排列的多个 Series 组成，每列可以有不同的数据类型（整数、浮点数、字符串等）。
- 同时具有行索引（index）和列索引（columns）。
- 支持丰富的数据操作，如数据选取、筛选、排序、分组、统计等。

在实际数据分析中，绝大部分数据处理工作都是围绕 DataFrame 展开的。

```python
# 创建一个简单的 DataFrame
data = {
    "姓名": ["小明", "小红", "小刚"],
    "年龄": [20, 21, 19],
    "成绩": [88, 92, 85]
}

df = pd.DataFrame(data)
print("创建的 DataFrame：")
print(df)
print()

# 查看行索引
print("行索引 (index)：", df.index)
print()

# 查看列索引
print("列索引 (columns)：", df.columns)
print()

# 查看单列（Series）
print("姓名列：")
print(df["姓名"])
print()

# 查看多列（DataFrame）
print("姓名和成绩列：")
print(df[["姓名", "成绩"]])
```

### 2.3 DataFrame 数据修改

在数据分析过程中，对 DataFrame 进行**增、删、改、查**是最基本也是最常见的操作。掌握这些操作，能够让你灵活地处理各种数据清洗与整理任务。

#### DataFrame 常见修改操作：

- **新增列**：直接赋值即可新增新列，列长度需与已有行数一致。
- **修改元素**：通过 `.loc[]` 或 `.iloc[]` 精确定位并修改指定元素。
- **删除列**：使用 `drop(columns=...)` 删除指定列。
- **删除行**：使用 `drop(index=...)` 删除指定行。

```python
import pandas as pd

# 创建一个简单的 DataFrame
data = {
    "姓名": ["小明", "小红", "小刚"],
    "年龄": [20, 21, 19],
    "成绩": [88, 92, 85]
}

df = pd.DataFrame(data)
print("原始 DataFrame：\n", df)
print()

# 新增列（比如添加"年级"列）
df["年级"] = ["大一", "大二", "大一"]
print("新增列后的 DataFrame：\n", df)
print()

# 修改单个元素（使用 loc，根据标签修改 —— 修改小明的成绩）
df.loc[0, "成绩"] = 90
print("使用 loc 修改成绩后的 DataFrame：\n", df)
print()

# 修改单个元素（使用 iloc，根据位置修改 —— 修改小红的年龄）
df.iloc[1, 1] = 22  # 第二行第二列（小红的年龄）改为22
print("使用 iloc 修改年龄后的 DataFrame：\n", df)
print()

# 删除列（删除"年级"列）
df = df.drop(columns=["年级"])
print("删除列后的 DataFrame：\n", df)
print()

# 删除行（删除小刚所在的行，索引为2）
df = df.drop(index=2)
print("删除行后的 DataFrame：\n", df)
```

### 3.3 数据选择与切片

数据选择与切片（Selection & Slicing）是 Pandas 中最基础也是最重要的技能之一。  
掌握灵活的数据选择方法，可以帮助我们快速定位、提取和处理感兴趣的数据子集。

在 Pandas 中，常用的数据选择方式包括：

- 通过列名选择单列或多列
- 通过条件表达式筛选行
- 使用 `.iloc[]` 按**位置**选择数据
- 使用 `.loc[]` 按**标签**选择数据

```python
# 选择单列（Series形式）
sepal_length = iris['sepal_length']
print("单列 sepal_length：")
print(sepal_length.head())
print()

# 选择多列（DataFrame形式）
subset = iris[['sepal_length', 'species']]
print("选择 sepal_length 和 species 两列：")
print(subset.head())
print()

# 根据条件筛选数据（sepal_length 大于 5.0）
large_sepal = iris[iris['sepal_length'] > 5.0]
print("筛选出 sepal_length > 5.0 的数据：")
print(large_sepal.head())
print()

# 使用 iloc 按位置选择（第1~3行，全部列）
print("使用 iloc 按位置选择第1到第3行：")
print(iris.iloc[0:3, :])
print()

# 使用 loc 按标签选择（选择前3行的 sepal_width 和 species列）
print("使用 loc 按标签选择前3行的 sepal_width 和 species列：")
print(iris.loc[0:2, ['sepal_width', 'species']])
```

### 3.5 数据排序与索引

在数据分析中，经常需要按照某列的数值进行排序，或者根据特定的列重新设定索引（Index），以便更方便地查询、分组或可视化。

Pandas 提供了灵活的数据排序与索引设置方法，包括：

- 使用 `sort_values()` 按列排序
- 使用 `set_index()` 重新设置行索引
- 使用 `reset_index()` 将索引还原为普通列

掌握这些方法可以帮助我们更高效地组织和操作数据。

```python
# 按sepal_length升序排列
iris_sorted = iris.sort_values(by="sepal_length")
print(iris_sorted.head())

# 按species列进行索引
iris_indexed = iris.set_index("species")
print(iris_indexed.head())

# 重置索引
iris_reset = iris_indexed.reset_index()
print(iris_reset.head())
```

### 3.6 分组与聚合操作

在数据分析中，经常需要按照某一列或多列进行分组，并对每个组应用聚合计算，比如求均值、最大值、最小值、数量统计等。  
这类操作在 Pandas 中可以通过 `.groupby()` 配合 `.mean()`、`.agg()` 等方法轻松实现。

分组聚合是数据探索分析（EDA）阶段非常重要的技能，能够帮助我们快速了解数据在不同类别下的分布特征。

```python
# 按 species 列进行分组，计算每组的均值
group_mean = iris.groupby('species').mean()
print("每种鸢尾花特征的均值：\n", group_mean)
print()

# 按 species 列进行分组，使用多种聚合函数
group_summary = iris.groupby('species').agg(['mean', 'max', 'min', 'count'])
print("每种鸢尾花特征的详细统计：\n", group_summary)
```
