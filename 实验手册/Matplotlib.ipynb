## Part 1. 什么是 matplotlib？

`Matplotlib` 是 Python 中最基础、最广泛使用的二维绘图库。它可以帮助我们以编程方式将数据以图形的形式呈现出来，包括折线图、柱状图、散点图、直方图、箱线图等多种常见图表形式。

### 1.1 简介

- `Matplotlib` 是一个开源的绘图库，最初由 John D. Hunter 开发。
- 它的子模块 `matplotlib.pyplot`（简称 `plt`）提供了类似 MATLAB 的绘图接口，简单易用。
- 它与 `NumPy` 和 `Pandas` 等数据分析工具无缝集成，适用于科研、工程和教学中的数据可视化任务。

### 1.2 为什么要学习 matplotlib？

- **易上手**：即使你从未接触过可视化，只需要几行代码就能画出标准图表。
- **高度灵活**：可以细致控制图像的每个部分，包括坐标轴、颜色、线条样式、字体等。
- **广泛兼容**：可与 Jupyter Notebook、Pandas、NumPy、Tkinter 等工具无缝配合使用。
- **可保存图像**：支持将图像保存为 PNG、SVG、PDF 等格式，方便报告或论文使用。

### 1.3 应用场景

- 教学演示
- 科学计算和数据分析
- 项目报告和论文图表绘制
- 产品数据可视化与监控仪表板基础

## Part 2. matplotlib.pyplot 接口基础教学

`matplotlib.pyplot`（简称 `plt`）是 `Matplotlib` 中最常用的绘图接口，提供了一组类似于 MATLAB 的命令式函数，用户可以通过这些函数快速创建图表并控制其外观。

### 2.1 常见绘图函数

| 函数名              | 功能说明                       | 常用参数                            |
|---------------------|--------------------------------|--------------------------------------|
| `plt.plot()`        | 绘制折线图                     | `x`, `y`, `label`, `color`, `linestyle`, `marker`, `linewidth` |
| `plt.scatter()`     | 绘制散点图                     | `x`, `y`, `color`, `marker`, `s`     |
| `plt.bar()`         | 绘制柱状图                     | `x`, `height`, `color`, `label`      |
| `plt.barh()`        | 绘制水平柱状图                 | `y`, `width`, `color`, `label`       |
| `plt.hist()`        | 绘制直方图                     | `x`, `bins`, `color`, `density`      |
| `plt.boxplot()`     | 绘制箱线图                     | `x`, `vert`, `patch_artist`          |

### 2.2 常见辅助函数

| 函数名               | 功能说明                           |
|----------------------|------------------------------------|
| `plt.title()`         | 设置图表标题                       |
| `plt.xlabel()`        | 设置 x 轴标签                      |
| `plt.ylabel()`        | 设置 y 轴标签                      |
| `plt.legend()`        | 显示图例（需设置 label 参数）      |
| `plt.grid()`          | 显示或关闭网格线                   |
| `plt.xlim()` / `plt.ylim()` | 设置坐标轴显示范围             |
| `plt.xticks()` / `plt.yticks()` | 设置坐标轴刻度值           |
| `plt.savefig()`       | 保存图像到文件                     |
| `plt.show()`          | 显示图表（必须调用才会显示图形）   |

### 2.3 样式设置方法

在 `plot()`、`scatter()` 等函数中，可以通过以下参数自定义图形样式：

- `color`：颜色（如 `'r'`, `'blue'`, `'#00ff00'`）
- `linestyle`：线型（如 `'-'`, `'--'`, `':'`, `'-.'`）
- `marker`：标记点样式（如 `'o'`, `'s'`, `'^'`, `'*'`）
- `linewidth`：线条宽度
- `markersize`：标记大小
- `label`：图例名称（配合 `plt.legend()` 使用）


````python
import matplotlib.pyplot as plt

# 准备数据
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

# 使用 plot 函数绘制折线图
# 参数：
#   x, y：分别为横轴和纵轴数据
plt.plot(x, y)

# 显示图形
plt.show()
````

````python
# 再次绘制一条折线图，并添加详细信息
plt.plot(x, y, label='Prime Growth')

# 设置标题
plt.title("Prime Numbers Line Chart")

# 设置坐标轴标签
plt.xlabel("X Axis")
plt.ylabel("Y Axis")

# 显示图例（label 参数来自 plot 函数）
plt.legend()

# 添加网格线，增强可读性
plt.grid(True)

# 显示图形
plt.show()
````

````python
# 自定义样式：红色圆点、虚线、线宽为2
plt.plot(x, y, color='red', linestyle='--', marker='o', linewidth=2, label='Styled Line')

plt.title("Styled Line Example")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.legend()
plt.grid(True)
plt.show()
````

````python
# 设置坐标轴范围
plt.plot(x, y, marker='s', label='Square Marker')
plt.title("Axis Limit Example")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")

# 设置坐标轴显示范围
plt.xlim(0, 6)  # 横轴范围为 0 到 6
plt.ylim(0, 12) # 纵轴范围为 0 到 12

# 自定义刻度值
plt.xticks([0, 1, 2, 3, 4, 5, 6])
plt.yticks([0, 2, 4, 6, 8, 10, 12])

plt.legend()
plt.grid(True)
plt.show()
````

## Part 3. 使用 Matplotlib 绘图的完整流程与图表类型概览

在学习了基础的函数与样式设置之后，我们进一步总结一下使用 Matplotlib（特别是 `plt` 接口）绘图的一般流程，帮助大家在实践中形成系统性思维。

### 3.1 Matplotlib 绘图的一般流程

1. **导入库**
   - 使用 `import matplotlib.pyplot as plt` 进行导入
2. **准备数据**
   - 可以使用列表、元组、NumPy 数组，或 Pandas 的 Series、DataFrame
3. **调用绘图函数**
   - 如 `plt.plot()`、`plt.bar()`、`plt.scatter()` 等
4. **添加图表细节**
   - 使用 `plt.title()`、`plt.xlabel()`、`plt.ylabel()` 添加标题与标签
   - 使用 `plt.legend()` 添加图例，`plt.grid()` 添加网格线
   - 使用 `plt.xlim()`、`plt.ylim()` 控制显示范围
5. **显示或保存图形**
   - 使用 `plt.show()` 显示图像
   - 使用 `plt.savefig("filename.png")` 将图像保存到本地

这一套流程可以适配各种图表类型，并通过灵活的参数控制图像样式，是掌握 Matplotlib 的关键。

### 3.2 Matplotlib 支持的常见图表类型

虽然 Matplotlib 功能非常强大，但在实际教学和应用中，以下几类图是最常用、最实用的：

| 图表类型   | 函数             | 适用场景                       |
|------------|------------------|--------------------------------|
| 折线图     | `plt.plot()`     | 显示连续变化趋势，如时间序列数据 |
| 散点图     | `plt.scatter()`  | 展示两个变量之间的相关关系       |
| 柱状图     | `plt.bar()`      | 对比各类数据值的大小             |
| 水平柱状图 | `plt.barh()`     | 与 `bar()` 类似，但用于横向展示   |
| 直方图     | `plt.hist()`     | 展示数据分布密度（统计频数）     |
| 箱线图     | `plt.boxplot()`  | 分析数据的分布与异常值情况       |

### Part 4.1 使用鸢尾花数据集绘制折线图

折线图通过将一系列数据点用线段连接起来，展示数据随横轴变化的趋势，常用于展示时间序列或有序数据的变化过程。

在本例中，我们将绘制不同品种的鸢尾花样本中，花萼长度（sepal length）的变化曲线，并通过颜色加以区分，以便进行对比观察。


````python
import pandas as pd
import matplotlib.pyplot as plt

# 使用 pandas 读取鸢尾花数据集
# 数据集包括 sepal length, sepal width, petal length, petal width 和 species（品种）
df = pd.read_csv('./data/iris.csv')

# 查看前5行数据，确认读取成功
print(df.head())

# 为每个类别分别绘制 sepal_length 的折线图
# 按照品种分类，遍历每个类别分别绘图
species_list = df['species'].unique()

# 使用 enumerate 给每一类设定一个偏移量，防止绘图时线重合
for i, species in enumerate(species_list):
    # 获取当前品种的数据
    sub_df = df[df['species'] == species]
    
    # 使用 plot 绘制折线图
    # 参数说明：
    #   x：样本编号（使用 sub_df 的索引）
    #   y：sepal_length 数值
    #   label：用于生成图例的品种名称
    #   marker：设置点的样式为圆点
    #   linestyle：使用虚线（可选）
    plt.plot(sub_df.index, sub_df['sepal_length'],
             label=species,
             marker='o',
             linestyle='--')

# 添加标题和坐标轴标签
plt.title("Sepal Length Line Plot by Species")
plt.xlabel("Sample Index")
plt.ylabel("Sepal Length (cm)")

# 添加图例和网格线
plt.legend()
plt.grid(True)

# 显示图形
plt.show()
````

### Part 4.2 使用鸢尾花数据集绘制散点图

散点图通过在坐标系中绘制一组独立的数据点，用于展示两个变量之间的关系，常用于判断变量之间是否存在相关性或聚类趋势。

在本例中，我们将以花萼长度（sepal length）为横轴，花萼宽度（sepal width）为纵轴，绘制不同品种鸢尾花的散点图，并通过不同颜色加以区分，以便进行直观比较。


````python
import pandas as pd
import matplotlib.pyplot as plt

# 读取数据
df = pd.read_csv('./data/iris.csv')

# 获取所有品种的唯一值
species_list = df['species'].unique()

# 为不同品种绘制散点图
for species in species_list:
    # 获取当前品种的子集数据
    sub_df = df[df['species'] == species]
    
    # 使用 scatter 绘制散点图
    # 参数说明：
    #   x：sepal_length，作为横坐标
    #   y：sepal_width，作为纵坐标
    #   label：品种名称，用于生成图例
    #   alpha：设置点的透明度，避免过于密集
    plt.scatter(sub_df['sepal_length'],
                sub_df['sepal_width'],
                label=species,
                alpha=0.8)

# 添加标题与坐标轴标签
plt.title("Sepal Length vs Sepal Width (Scatter Plot)")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")

# 添加图例与网格线
plt.legend()
plt.grid(True)

# 显示图形
plt.show()
````

### Part 4.3 使用鸢尾花数据集绘制柱状图和水平柱状图

柱状图通过垂直条形的高度展示每个类别的数值大小，适用于展示分类变量的比较；而水平柱状图则将条形旋转为水平方向，在标签较长或空间受限时更易阅读。

在本例中，我们将绘制每个品种鸢尾花的平均花瓣长度（petal length），分别用柱状图和水平柱状图展示其对比效果。


````python
import pandas as pd
import matplotlib.pyplot as plt

# 读取数据
df = pd.read_csv('./data/iris.csv')

# 计算每个品种的平均 petal_length
mean_petal_length = df.groupby('species')['petal_length'].mean()


# 绘制柱状图
plt.figure(figsize=(8, 6))  # 设置图像大小

bars = plt.bar(mean_petal_length.index, mean_petal_length.values,
               color='#69b3a2', edgecolor='black')

plt.title("Average Petal Length by Species")
plt.xlabel("Species")
plt.ylabel("Petal Length (cm)")

# 在每个柱子上方添加具体数值标签
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width() / 2, height + 0.05,
             f'{height:.2f}', ha='center', va='bottom')

plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()


# 绘制水平柱状图
plt.figure(figsize=(8, 6))

bars = plt.barh(mean_petal_length.index, mean_petal_length.values,
                color='#ffcc5c', edgecolor='black')

plt.title("Average Petal Length by Species (Horizontal)")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Species")

# 在每个条形末尾添加数值标签
for bar in bars:
    width = bar.get_width()
    plt.text(width + 0.05, bar.get_y() + bar.get_height() / 2,
             f'{width:.2f}', va='center')

plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
````
